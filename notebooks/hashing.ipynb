## Cryptonic hash

```python
import hashlib
```

```python
def sha256_image(image_file):
    with open(image_file, "rb") as f:
        return hashlib.sha256(f.read()).hexdigest()
```

```python
for image in image_paths:
    hash = sha256_image(image)
    print(hash)
```

```python
from PIL import Image
from imagehash import average_hash, phash, phash_simple, dhash, dhash_vertical, whash, colorhash, crop_resistant_hash, hex_to_hash, hex_to_multihash
```

```python
HASH_METHODS = {
    "average": average_hash,
    "phash": phash,
    "phash_simple": phash_simple,
    "dhash": dhash,
    "dhash_vertical": dhash_vertical,
    "whash": whash,
    "colorhash": colorhash,
    "crop_resistant": crop_resistant_hash,
}
```

```python
def image_hash(image_file, method="phash"):
    """
    Generate an image hash using a selected imagehash method.

    Args:
        image_file (str): Path to the input image.
        method (str, optional):Hashing method to use. Defaults to "phash". Supported methods are:
                "average"
                "phash"
                "phash_simple"
                "dhash"
                "dhash_vertical"
                "whash"
                "colorhash"
                "crop_resistant"

    Returns:
        ImageHash: Hash object returned by the selected imagehash  algorithm.
    """

    if method not in HASH_METHODS:
        raise ValueError(f"Unknown hash method: {method}. " f"Choose from: {list(HASH_METHODS.keys())}")

    image = Image.open(image_file)

    hash_function = HASH_METHODS[method]

    return hash_function(image)

def all_image_hashes(image_file):
    """
    Generate all supported perceptual hashes for an image.
    Supported methods are:
                "average"
                "phash"
                "phash_simple"
                "dhash"
                "dhash_vertical"
                "whash"
                "colorhash"
                "crop_resistant"

    Args:
        image_file (str): Path to the input image.

    Returns:
        dict: Dictionary mapping each hash method name to its resulting hash stored as string.
    """
    return {
        name: str(image_hash(image_file, name))
        for name in HASH_METHODS.keys()
    }


def restore_hash(hash_value):
    """
    Convert a hexadecimal (multi-)hash back into ImageHash objects.

    Args:
        hash_value (str): Hexadecimal representation of a (multi-)hash.

    Returns:
        list: A list of ImageHash objects reconstructed from the hexadecimal (multi-)hash.
    """

    return hex_to_multihash(hash_value)
```

## Example usage

```python
hashes = []
for image_file in image_paths:
    image_hashes = {
        "image_id": image_file,
        "hashes": all_image_hashes(image_file),
        "sha256": sha256_image(image_file)
    }

    hashes.append(image_hashes)

display(hashes)

restore_hash(hashes[0]['hashes']['phash'])
```
